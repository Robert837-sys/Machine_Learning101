## Data Types
- We can split the data types into 3 main categories:
- ***Numerical***
- ***Categorical***
- ***Ordinal***

**Numerical** data are numbers, and can be split into 2 main categories:

-***Discrete Data*** -counted data that are limited to integers. Example: The number of cars passing by.

-***Continuous Data*** -measured data that can be any number. Example: The price of an item, or the size of an item.

**Categorical** data are values that can't be measured up against each other. E.g., a color value, or any yes/no values.

**Ordinal** data are like categorical data, but can be ,measured up against each other. E.g., school grades where A is better than B and so on.

```python
# Binomial Distribution 

x=np.random.binomial(n=10,p=0.5,size=1000)
plt.hist(x,bins=np.arange(-0.5,11.5,1),edgecolor='black')
plt.title('Binomial Distribution')
plt.xlabel('Number of Heads')
plt.ylabel('Frequency')
plt.show()
```

```python
# Poisson Distribution
# A poisson distribution describes how many times an event occurs 
#within a fixed interval of time, space, or another measurement.
# lam=5: Expected number of events per interval
# size=1000 number of intervals simulated 
np.random.seed(42)
x=np.random.poisson(lam=5,size=1000)
plt.hist(x,bins=np.arange(-0.5,16.5,1),edgecolor='black')
plt.title('Poisson Distribution')
plt.xlabel('Number of Events')
plt.ylabel('Frequency')
plt.show()
```

```python
# An exponential distribution describes the waiting time between consecutive
#events in a Poisson process. 
# scale=5: Average waiting time.
# size=1000: Number of waiting-time observations.
np.random.seed(42)
x=np.random.exponential(scale=5,size=1000)
plt.hist(x,bins=30,edgecolor='black')
plt.title('Exponential Distribution')
plt.xlabel('Waiting Time')
plt.ylabel('Frequency')
plt.show()
```

# Log-Normal Distribution
A log-normal distribution occurs when the logarithm of a variable follows a normal distribution.
In simple terms, the original data is not normally distributed, but its logarithmic values are.

```python
x=np.random.lognormal(
    mean=2,
    sigma=0.5,
    size=1000
)
plt.hist(x,bins=30,edgecolor='black')
plt.title('Log-Normal Distribution')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()
```

# Geometric Distribution
A geometric distribution describes the number of independent trials required to obtain the first success.


```python
np.random.seed(42)
x=np.random.geometric(p=0.3,size=1000)
plt.hist(x,bins=np.arange(0.5,20.5,1),edgecolor='black')
plt.title('Geometric Distribution')
plt.xlabel('Trials Before First Success')
plt.ylabel('Frequency')
plt.show()
```

# Beta Distribution
A beta distribution models probabilities and other values bounded between 0 and 1.
It's particularly useful when analysing proportions, probabilities, and conversion rates.

```python
np.random.seed(42)
x=np.random.beta(
    a=2,
    b=5,
    size=1000
)
# a=2: First shape parameter
# b=5: second shape parameter
# size = 1000: Number of observations.
plt.hist(x, bins=30, edgecolor='black')

plt.title('Beta Distribution')
plt.xlabel('Probability')
plt.ylabel('Frequency')
plt.show()
```

```python
from scipy import stats
x = [5,7,8,7,2,17,2,9,4,11,12,9,6]
y = [99,86,87,88,111,86,103,87,94,78,77,85,86]

slope,intercept,r,p,std_err=stats.linregress(x,y)

def myfunc(x):
    return slope * x+intercept

mymodel=list(map(myfunc,x))

plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
```

r-: Correlation coefficient (-1 to 1): strength and direction of the linear relationship

p-: value testing whether the slope is really different from 0 (small, e.g. < 0.05, means a statistically significant relationship)

std_err: Standard error of the slope: how uncertain the slope estimate is
