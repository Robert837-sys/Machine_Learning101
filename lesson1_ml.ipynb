## Data Types
- We can split the data types into 3 main categories:
- ***Numerical***
- ***Categorical***
- ***Ordinal***

**Numerical** data are numbers, and can be split into 2 main categories:

-***Discrete Data*** -counted data that are limited to integers. Example: The number of cars passing by.

-***Continuous Data*** -measured data that can be any number. Exmple: The price of an item, or the size of an item.

**Categorical** data are values that can't be measured up against each other. E.g., a color valu, or any yes/no values.

**Ordinal** data are like categorical data, but can be ,measured up against each other. E.g., school grades where A is better than B and so on.

```python
import numpy as np
speed=[99,86,87,88,111,86,103,87,94,78,77,85,86]
x=np.mean(speed)
print(x)
```

```python
y=np.median(speed)
y
```

```python
from scipy import stats
speed=[99,86,87,88,111,86,103,87,94,78,77,85,86]
x=stats.mode(speed)
x
```

```python
speed=[86,87,88,86,87,85,86]
x=np.std(speed)
print(x)
```

```python
speed=[32,111,138,28,59,77,97]
x=np.var(speed)
print(x)
```

# ML Percentiles

- Percentiles are used in statistics to give you a number that describes the value that a given percent of the values are lower than.

```python
ages=[5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]
```

```python
x=np.percentile(ages,75)
print(x)
```

```python
x=np.percentile(ages,90)
print(x)
```

# Random Data Distributions
In Ml learning the data sets can contain thousands- or even millions of values.
You might not have real world data when you're testing an anlgorithm, you might have to use randomly generated values.
Numpy module can help us with that!

```python
x=np.random.normal(5.0,1.0,1000)
y=np.random.normal(10,2.0,1000)

plt.scatter(x,y)
plt.show()
```
